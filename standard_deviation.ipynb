```python
import random
import math

data = [random.randint(10, 100) for _ in range(5)]
print("Dataset:", data)
```

```python
mean = sum(data) / len(data)
print("Mean:", mean)
```

```python
variance = sum((x - mean) ** 2 for x in data) / len(data)
std_dev = math.sqrt(variance)
print("Standard Deviation:", std_dev)
```

```python
z_scores = [(x - mean) / std_dev for x in data]
print("Z-scores:", z_scores)
```

```python
#finding standard devaition using scikitlearn

#it is used to generate a datset of random integer between 1 and 100000

import numpy as np 
import pandas as pd

np.random.seed(45)
X = np.random.randint(1,100000,size = 5)
print ("Dataset:", X)
```

```python
from sklearn.preprocessing import StandardScaler    
s = StandardScaler()
X_scaled = s.fit_transform(X.reshape(-1, 1))

#using this print we get ouput as this 
print(X_scaled)
# [[-0.50880213]
#  [-1.01104002]
#  [ 1.54336175]
#  [ 0.80575974]
#  [-0.82927934]]


#using this print we get ouput as this 
#print( X_scaled.flatten())
#[-0.50880213 -1.01104002  1.54336175  0.80575974 -0.82927934]

```

```python
mean = np.mean(X_scaled)
print(mean)
```

```python
# standard deviation = np.std(X_scaled)
# print(standard deviation)

sq = 0
for i in X_scaled:
    #sq += (i - mean) ** 2
    sq = sq + ((i - mean) ** 2)
    std = (sq/len(X_scaled)) ** 0.5
print(std)
```

```python
X_scaled = []
for i in X_scaled:
    X_scaled.append((i - mean) / std)
print(X_scaled)
```
